# Doanh thu

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
#Đọc file
csv_file_path = r'D:\DATN\DATN---Phantom-Data\Data dự án\data\cleaned\superstore_cleaned.csv'
df = pd.read_csv(csv_file_path)

print("\nDataFrame from CSV file:")
print(df)

#In ra 5 hàng đầu tiên của DataFrame df để khám phá dữ liệu.
print("\nFirst 5 rows of the DataFrame:")
print(df.head())

#Sử dụng phương thức info() để hiển thị thông tin về cấu trúc của DataFrame df, bao gồm số lượng hàng và cột, và kiểu dữ liệu của các cột.
print("\nDataFrame info:")
print(df.info())
```

# Doanh thu theo năm

```python
sales_year = df.groupby("order_year")["sales"].sum()

plt.figure(figsize=(8,5))
plt.plot(sales_year.index,
         sales_year.values,
         marker='o',
         linewidth=3)

plt.title("Revenue by Year")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

# Top 10 quốc gia doanh thu cao nhất

```python
country_sales = (
    df.groupby("country")["sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))
country_sales.plot(kind="bar")

plt.title("Top 10 Countries by Sales")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

# Doanh thu theo Category

```python
category_sales = df.groupby("category")["sales"].sum()

plt.figure(figsize=(6,5))

category_sales.plot(
    kind="bar",
    color=["skyblue","orange","green"]
)

plt.title("Sales by Category")
plt.ylabel("Sales")

plt.show()
```

# Lợi nhuận theo Category

```python
category_profit = df.groupby("category")["profit"].sum()

plt.figure(figsize=(6,5))

category_profit.plot(
    kind="bar",
    color=["green","red","blue"]
)

plt.title("Profit by Category")
plt.ylabel("Profit")

plt.show()
```

# Top 10 Sub Category

```python
sub_profit = (
    df.groupby("sub_category")["profit"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,5))

sub_profit.plot(kind="bar")

plt.title("Top 10 Sub Categories")
plt.ylabel("Profit")

plt.show()
```

# Doanh thu theo Segment

```python
segment_sales = df.groupby("segment")["sales"].sum()

plt.figure(figsize=(6,6))

plt.pie(
    segment_sales,
    labels=segment_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales by Segment")

plt.show()
```

# Discount vs Profit

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["discount"],
    df["profit"],
    alpha=0.3
)

plt.xlabel("Discount")
plt.ylabel("Profit")
plt.title("Discount vs Profit")

plt.show()
```

# Doanh thu theo Market

```python
market_sales = df.groupby("market")["sales"].sum()

plt.figure(figsize=(8,5))

market_sales.plot(
    kind="bar",
    color="steelblue"
)

plt.title("Sales by Market")
plt.ylabel("Sales")

plt.show()
```

# Profit theo Ship Mode

```python
ship_profit = df.groupby("ship_mode")["profit"].sum()

plt.figure(figsize=(7,5))

ship_profit.plot(
    kind="bar",
    color="green"
)

plt.title("Profit by Ship Mode")

plt.show()
```
